```python
from astroquery.mast import Observations
```

```python
import numpy as np
```

```python
tic = 219796019
```

```python
obs = Observations.query_criteria(objectname=f"TIC {tic}"
                                      ,radius=0
                                      ,obs_collection=["TESS", "HLSP"]
                                      ,provenance_name=["SPOC", "TESS-SPOC"]
                                      ,target_name=[f"{tic}", "TESS FFI"]
                                      )
```

```python
obs[np.bitwise_and(obs['provenance_name']=='SPOC', obs['target_name']==f"{tic}")]
```

```python
obs[obs['provenance_name'] == "TESS-SPOC"]['sequence_number']
```
