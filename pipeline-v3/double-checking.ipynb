```python
import lightkurve as lk
%matplotlib inline
```

```python
lcs = lk.search_lightcurve("TIC 219033887", mission="TESS", author="SPOC").download_all()
```

```python
lcs.stitch().scatter()
```

```python
pg = lcs.stitch().to_periodogram(max_period=5)
```

```python
lcs.stitch().fold(pg.period_at_max_power).scatter()
```

```python
pg.period_at_max_power
```

```python
lc2 = lk.search_lightcurve("TIC 11688264", mission="TESS", author="SPOC").download_all().stitch()
lc2.plot()
```

```python
pg = lc2.to_periodogram(maximum_period=10)
```
